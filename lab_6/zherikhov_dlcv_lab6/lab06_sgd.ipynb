## 3. Random search baseline

The cell below evaluates 20 independently sampled weight matrices.
It keeps the one with the smallest training loss. Gradient descent
will instead use the local slope to choose every update.


```python
data = dlcv2026.make_toy_classification(seed=11)

plt.scatter(
    data["X_train"][:, 0],
    data["X_train"][:, 1],
    c=data["y_train"],
    cmap="viridis",
    s=16,
)
plt.title("Training data")
plt.xlabel("feature 1")
plt.ylabel("feature 2")
plt.show()

generator = torch.Generator().manual_seed(4)
best_loss = float("inf")
best_random_W = None
for _ in range(20):
    candidate_W = 0.5 * torch.randn(3, 3, generator=generator)
    candidate_loss, _ = linear_classifier.svm_loss_vectorized(
        candidate_W,
        data["X_train"],
        data["y_train"],
        reg=1e-3,
    )
    if float(candidate_loss) < best_loss:
        best_loss = float(candidate_loss)
        best_random_W = candidate_W

random_predictions = linear_classifier.predict_linear_classifier(
    best_random_W, data["X_val"]
)
random_accuracy = linear_classifier.accuracy(
    random_predictions, data["y_val"]
)
print(f"best random-search loss: {best_loss:.4f}")
print(f"random-search validation accuracy: {100 * random_accuracy:.2f}%")
```

## 6. Full-batch gradient descent and minibatch SGD

Full-batch gradient descent uses every training example for each
update. Its curve is smooth but each step uses more data. Minibatch
SGD uses fewer examples per step, so its loss is noisier.


```python
generator = torch.Generator().manual_seed(5)
W_full = 1e-3 * torch.randn(3, 3, generator=generator)
full_batch_history = []
for _ in range(100):
    loss, gradient = linear_classifier.svm_loss_vectorized(
        W_full,
        data["X_train"],
        data["y_train"],
        reg=1e-3,
    )
    full_batch_history.append(float(loss))
    W_full = W_full - 0.08 * gradient

plt.plot(full_batch_history, label="full batch")
plt.plot(minibatch_history[:100], alpha=0.75, label="minibatch")
plt.xlabel("iteration")
plt.ylabel("loss")
plt.title("Full-batch and minibatch optimization")
plt.legend()
plt.show()
```

## 7. Learning-rate experiment

Run the same training procedure with four learning rates. Compare
how quickly the curves fall and how much they fluctuate.


```python
learning_rates = [1e-4, 5e-3, 8e-2, 2.0]
for learning_rate in learning_rates:
    _, history = linear_classifier.train_linear_classifier(
        linear_classifier.svm_loss_vectorized,
        data["X_train"],
        data["y_train"],
        learning_rate=learning_rate,
        reg=1e-3,
        num_iters=200,
        batch_size=32,
        seed=5,
    )
    plt.plot(history, label=f"lr={learning_rate:g}")

plt.ylim(0, 2.1)
plt.xlabel("iteration")
plt.ylabel("minibatch loss")
plt.title("Effect of the learning rate")
plt.legend()
plt.show()
```

## 8. Wrap-up

Answer briefly in a new Markdown cell:

1. Why does the negative gradient point toward the steepest local decrease?
2. Why is a minibatch loss curve noisier than a full-batch curve?
3. What can happen when the learning rate is too small or too large?


### Answers

**1. Why does the negative gradient point toward the steepest local decrease?**

Near a point $W$, the loss is well approximated by its first-order Taylor expansion: $L(W + \varepsilon u) \approx L(W) + \varepsilon\, \nabla_W L \cdot u$. For a unit direction $u$, the dot product $\nabla_W L \cdot u = \|\nabla_W L\|\cos\theta$ is smallest (equal to $-\|\nabla_W L\|$) when $\theta = 180^\circ$, i.e. when $u = -\nabla_W L / \|\nabla_W L\|$. So a small step along the negative gradient decreases the loss faster than a step of the same size in any other direction. This is a local property: it only holds for sufficiently small steps.

**2. Why is a minibatch loss curve noisier than a full-batch curve?**

At each iteration, SGD computes the loss and gradient on a random subset (64 or 32 examples) instead of all 360 training examples. This gives an unbiased but noisy estimate of the full gradient: each batch contains a different mix of easy and hard examples, so both the loss value and the update direction vary from step to step. Full-batch gradient descent evaluates the same deterministic objective at every step, so its curve is smooth (see the plot in Section 6). The smaller the batch, the higher the variance of the estimate and the noisier the curve.

**3. What can happen when the learning rate is too small or too large?**

- **Too small** (`lr=1e-4`): the steps are tiny, and after 200 iterations the loss only drops from about 2.0 to about 1.67. Training is very slow and may not converge within the iteration budget. With `lr=5e-3` the loss does converge, but noticeably slower than with `lr=0.08`.
- **Too large** (`lr=2`): the steps overshoot the minimum. The weights change abruptly, the curve shows spikes, and the loss stays higher than with `lr=0.08`. On this simple, nearly linearly separable dataset the SVM loss does not diverge, but in general a learning rate that is too large causes oscillations or divergence (the loss grows and may become `inf`/`nan`).
- A well-chosen value (here `lr=0.08`) gives a fast and stable decrease in the loss.
